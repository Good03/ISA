```python
import pandas as pd

data = pd.read_csv("data/NASDAQ100_Historical_Data.csv")

data.isnull().sum()

data.head()
data.shape
data.columns
data.info()
data.describe()
```

```python
data["Date"] = pd.to_datetime(data["Date"])
data = data.sort_values("Date")

data.info()
```

```python
import matplotlib.pyplot as plt

apple = data[data["Ticker"] == "AAPL"]

plt.figure(figsize=(10,5))
plt.plot(apple["Date"], apple["Close"])
plt.title("Apple Price Over Time")
plt.show()
```

```python
import matplotlib.pyplot as plt

adobe = data[data["Ticker"] == "ADBE"]

plt.figure(figsize=(10, 5))
plt.plot(adobe["Date"], adobe["Close"])
plt.title("Apple Price Over Time")
plt.show()
```
